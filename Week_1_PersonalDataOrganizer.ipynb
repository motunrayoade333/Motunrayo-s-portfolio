```python
#Personal Information (Example data(safe to share))
first_name = "Alex"
last_name = "Johnson"
age = 28
email = "alexjohnson@example.com"
city = "Toronto"
country = "Canada"

print ("===== PERSONAL PROFILE =====")
print (f"Name: {first_name} {last_name}")
print (f"Age: {age}")
print (f"Email: {email}")
print (f"Location: {city}, {country}")
```

```python
#How old will Alex be in 10 years?
age_in_years = 10
print ("Alex will be " + str(age + age_in_years) + " years old in 10 years.")
```

```python
# How old was Alex 5 years ago?
five_years_ago = 5
print("Alex was " + str(age - five_years_ago) + " years old 5 years ago.")
```

```python
# How many days has Alex been alive? (rough estimate)
days_alive = age * 365
print ("Alex has been alive for" + " " + str(days_alive) + " " + "days.")
```

```python
# How many years until retirement at 65?
years_until_retirement = 65 - age
print("Alex has" + " " + str(years_until_retirement) + " " + "years before retirement")
```

```python
#Identifying the data types of each value in the variable
print(type(first_name))
print(type(last_name))
print(type(age))
print(type(email))
print(type(city))
print(type(country))

""" Why did we use str(age) in print statements? What would happen if you removed it?
We used `str(age)` to convert the age from an integer (number) to a string (text) so it could be printed alongside other text.

If we removed it, Python would return a `TypeError` because it cannot directly combine a string and an integer."""
```
