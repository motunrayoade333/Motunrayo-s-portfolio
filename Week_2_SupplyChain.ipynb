```python
'''Write a function called approval_message(order_value) that returns:
•"Needs manager approval" if the value is above 100,000
• "Needs supervisor approval" if it is above 50,000
• "Auto-approved" otherwise'''

def approval_message(order_value):
    if order_value > 100000:
        return "Needs manager's approval"
    elif order_value > 50000:
        return "Needs supervisor's approval"
    else:
        return "Auto-approved"

print(approval_message(72500))
```

```python
# Task 2: Process multiple orders
order_values = [12000, 55000, 150000, 49999, 100001, 8000]

for value in order_values:
    print(value, "->", approval_message(value))

```

```python
# Task 3: Supplier status function
def check_supplier_status(supplier_name, on_time_rate):
    if on_time_rate >= 0.90:
        return f"{supplier_name}: Reliable"
    elif on_time_rate >= 0.80:
        return f"{supplier_name}: Monitor"
    else:
        return f"{supplier_name}: At Risk"

# Test with three suppliers
print(check_supplier_status("SupplierA", 0.94))
print(check_supplier_status("SupplierB", 0.86))
print(check_supplier_status("SupplierE", 0.74))
```

```python
import csv

with open("supply_chain_dataset.csv", newline="") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

   # Display column names by getting the keys of the first dictionary row
print("Column names:", list(rows[0].keys()))

# Display the number of data rows
print("Number of data rows:", len(rows))

# Display the first 5 rows
print("First 5 rows:")
for row in rows[:5]:
    print(row)
```

```python
# Task 5: Explore the dataset

# Question 1: How many rows of data are there?
print("Number of rows:", len(rows))

# Question 2: How many different supplier names are written in the file?
unique_suppliers = []
for row in rows:
    if row["supplier_name"] not in unique_suppliers:
        unique_suppliers.append(row["supplier_name"])

print("Different supplier names:", len(unique_suppliers))
print(unique_suppliers)

# Question 3: Highest total_cost (convert text to a number with float)
highest = None
for row in rows:
    if row["total_cost"] != "":
        cost = float(row["total_cost"])
        if highest is None or cost > highest:
            highest = cost

print("Highest total_cost:", highest)

# Question 4: Lowest total_cost
lowest = None
for row in rows:
    if row["total_cost"] != "":
        cost = float(row["total_cost"])
        if lowest is None or cost < lowest:
            lowest = cost

print("Lowest total_cost:", lowest)

# Question 5: How many orders have delivery_status "Delivered"?
delivered_count = 0
for row in rows:
    if row["delivery_status"] == "Delivered":
        delivered_count += 1

print("Orders with status Delivered:", delivered_count)

```

```python
# Challenge 1: Approval level based on order value AND supplier on-time rate
def approval_with_supplier(order_value, on_time_rate):
    if order_value > 100000 and on_time_rate < 0.80:
        return "Rejected: high value order from an at-risk supplier"
    elif order_value > 100000:
        return "Needs manager's approval"
    elif order_value > 50000 and on_time_rate < 0.80:
        return "Needs manager's approval: supplier is at risk"
    elif order_value > 50000:
        return "Needs supervisor's approval"
    else:
        return "Auto-approved"

# Tests
print(approval_with_supplier(150000, 0.74))
print(approval_with_supplier(150000, 0.94))
print(approval_with_supplier(72500, 0.86))
print(approval_with_supplier(8000, 0.74))
```

```python
# Challenge 2: Find all SupplierA orders, however the name is written
supplier_a_count = 0
supplier_a_total = 0

for row in rows:
    cleaned_name = row["supplier_name"].lower().replace(" ", "")
    if cleaned_name == "suppliera":
        supplier_a_count += 1
        if row["total_cost"] != "":
            supplier_a_total += float(row["total_cost"])

print("SupplierA orders:", supplier_a_count)
print("SupplierA total cost:", round(supplier_a_total, 2))
```

```python
# Challenge 3: Find which columns have empty values, and how many
empty_counts = {}

for column in rows[0].keys():
    empty_counts[column] = 0

for row in rows:
    for column in row:
        if row[column] == "":
            empty_counts[column] += 1

for column, count in empty_counts.items():
    if count > 0:
        print(column, ":", count, "empty values")
```
